```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime
```

```python
url = "https://r.jina.ai/https://www.ccny.cuny.edu/registrar/fall"

r = requests.get(url)

print(r.status_code)
```

```python
start = r.text.find("August 01")

print(start)
```

```python
lines = r.text.splitlines()

#beginning of table
beginning = next(i for i, line in enumerate(lines) if "August 01" in line)

#end of table
end = next(i for i, line in enumerate(lines[beginning:], start) if "January 1, 2022" in line)

#only calendar rows
calendar_lines = lines[beginning:end + 1]

for line in calendar_lines:
    print(line)
```

```python
data = []

for line in calendar_lines:
    line = line.replace("**", "").strip()

    if "|" not in line:
        continue

    #remove | 
    line = line.strip("|")

    #columns
    parts = [part.strip() for part in line.split("|")]

    if len(parts) >= 3:
        date_text = parts[0]
        dow = parts[1]
        text = parts[2]

        data.append([date_text, dow, text])

df = pd.DataFrame(data, columns=["date", "day", "text"])

df
```
